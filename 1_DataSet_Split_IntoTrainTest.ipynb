```python
import os
import shutil
from sklearn.model_selection import train_test_split

# ====== INPUT DATASET ======
data_dir = r"C:\Users\Karim\Contacts\Desktop\ACRIMA_N_DS"   # your original dataset

# ====== OUTPUT FOLDER ======
output_dir = r"C:\Users\Karim\Contacts\Desktop\split_data"

# create main folders
train_dir = os.path.join(output_dir, "train")
test_dir = os.path.join(output_dir, "test")

# create class folders inside train/test
class_names = os.listdir(data_dir)

for split in ["train", "test"]:
    for class_name in class_names:
        os.makedirs(os.path.join(output_dir, split, class_name), exist_ok=True)

# ====== COLLECT ALL FILE PATHS ======
image_paths = []
labels = []

for class_name in class_names:
    class_folder = os.path.join(data_dir, class_name)
    
    for img_name in os.listdir(class_folder):
        image_paths.append(os.path.join(class_folder, img_name))
        labels.append(class_name)

# ====== SPLIT ======
X_train, X_test, y_train, y_test = train_test_split(
    image_paths,
    labels,
    test_size=0.2,
    random_state=42,
    stratify=labels
)

# ====== COPY FUNCTION ======
def copy_images(X, y, split_name):
    for img_path, label in zip(X, y):
        filename = os.path.basename(img_path)
        dest_path = os.path.join(output_dir, split_name, label, filename)
        shutil.copy(img_path, dest_path)

# copy files
copy_images(X_train, y_train, "train")
copy_images(X_test, y_test, "test")

print("✅ Train/Test folders created successfully!")
```
