```python
class_names = ["Glaucoma","Normal"]
```

```python
X_train=Fused_Train_features
X_test=Fused_Test_features
y_train=labels_Train_encoded
y_test=labels_Test_encoded
```

```python
# SVM classifier with linear kernel
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

svm_classifier_lin = SVC(kernel='linear')
svm_classifier_lin.fit(X_train, y_train)
svm_pred_lin = svm_classifier_lin.predict(X_test)
svm_accuracy_lin = accuracy_score(y_test, svm_pred_lin)
print(f"SVM(linear kernel) Classifier Accuracy: {svm_accuracy_lin}")
```
