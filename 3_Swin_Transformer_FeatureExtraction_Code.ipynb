```python
def extract_features(folder_path):
    features_list = []
    labels = []

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(device)
    
    model = timm.create_model(
        'swin_base_patch4_window7_224',
        pretrained=True,
        num_classes=0  # IMPORTANT → removes classifier head
    )

    classes = sorted(os.listdir(folder_path))

    with torch.no_grad():
        for class_name in classes:
            class_path = os.path.join(folder_path, class_name)

            for img_name in os.listdir(class_path):
                img_path = os.path.join(class_path, img_name)

                img = cv2.imread(img_path)
                if img is None:
                    continue

                img = cv2.resize(img, (224, 224))
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

                img = img / 255.0
                img = torch.tensor(img, dtype=torch.float32).permute(2, 0, 1)

                img = (img - torch.tensor([0.485,0.456,0.406]).view(3,1,1)) / \
                      torch.tensor([0.229,0.224,0.225]).view(3,1,1)

                img = img.unsqueeze(0).to(device)

                feat = model(img)

                features_list.append(feat.cpu().numpy().squeeze())
                labels.append(class_name)

    return np.array(features_list), np.array(labels)
```

```python
import os
import cv2
import numpy as np
import torch
import timm

X_train, y_train = extract_features(r"C:\Users\Karim\Contacts\Desktop\split_data\train_AUGMENTED")
X_test, y_test = extract_features(r"C:\Users\Karim\Contacts\Desktop\split_data\test")

print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```

```python
import numpy as np

dataset_name = "AC"

# Save TRAIN
np.save(f"{dataset_name}_X_train_swin.npy", X_train)
np.save(f"{dataset_name}_y_train_swin.npy", y_train)

# Save TEST
np.save(f"{dataset_name}_X_test_swin.npy", X_test)
np.save(f"{dataset_name}_y_test_swin.npy", y_test)

print("✅ Train and Test features saved successfully!")
```
